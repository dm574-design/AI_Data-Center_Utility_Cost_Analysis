```python
import pandas as pd

# Load the EIA Excel workbook
file_path = "../data/raw/Sales_Ult_Cust_2025.xlsx"

# Display worksheet names
excel_file = pd.ExcelFile(file_path)

print(excel_file.sheet_names)
```

```python
# Load the States worksheet
df = pd.read_excel(file_path, sheet_name="States")

# Display the first 10 rows
display(df.head(10))

# Show the column names
print(df.columns.tolist())

# Show the dataset dimensions
print("Rows and columns:", df.shape)
```

```python
# Load the worksheet using the correct header rows
df = pd.read_excel(
    file_path,
    sheet_name="States",
    header=[1, 2]
)

# Display the first five rows
display(df.head())

# Check the column names
print(df.columns.tolist())

# Check dataset size
print("Rows and columns:", df.shape)
```

```python
# Display each column with its position
for i, col in enumerate(df.columns):
    print(i, col)
```

```python
# Select the identifying columns and first data group
selected_columns = df.iloc[:, [2, 6, 9, 10, 11]]

# Display the first 10 records
display(selected_columns.head(10))
```

```python
# Create a simplified dataset
electricity = selected_columns.copy()

# Rename columns
electricity.columns = [
    "Utility",
    "State",
    "Revenue_Thousand_Dollars",
    "Sales_MWh",
    "Customers"
]

# Convert numeric columns
numeric_cols = [
    "Revenue_Thousand_Dollars",
    "Sales_MWh",
    "Customers"
]

for col in numeric_cols:
    electricity[col] = pd.to_numeric(
        electricity[col], errors="coerce"
    )

# Calculate average electricity price in cents per kWh
electricity["Price_Cents_kWh"] = (
    electricity["Revenue_Thousand_Dollars"]
    / electricity["Sales_MWh"]
) * 100

# Display the first 10 rows
display(electricity.head(10))
```

```python
# Check the size of the dataset
print("Dataset size:", electricity.shape)

# Check missing values
print("\nMissing values:")
print(electricity.isnull().sum())

# Check for zero or negative electricity sales
print("\nZero or negative sales:")
print((electricity["Sales_MWh"] <= 0).sum())

# Display summary statistics
display(electricity.describe())
```

```python
import numpy as np

# Keep the original dataset unchanged
electricity_clean = electricity.copy()

# Replace infinity with missing values
electricity_clean = electricity_clean.replace(
    [np.inf, -np.inf], np.nan
)

# Keep records with positive sales and nonnegative revenue
electricity_clean = electricity_clean[
    (electricity_clean["Sales_MWh"] > 0) &
    (electricity_clean["Revenue_Thousand_Dollars"] >= 0)
].copy()

# Remove missing prices and identifiers
electricity_clean = electricity_clean.dropna(
    subset=["Utility", "State", "Price_Cents_kWh"]
)

# Check the cleaned dataset
print("Original records:", len(electricity))
print("Cleaned records:", len(electricity_clean))
print("Records excluded:", len(electricity) - len(electricity_clean))

display(electricity_clean.head(10))
```

```python
# Read the original worksheet header rows
header_preview = pd.read_excel(
    file_path,
    sheet_name="States",
    header=None,
    nrows=3
)

# Display the headers across all columns
for col in range(header_preview.shape[1]):
    print(
        col,
        "| Group:", header_preview.iloc[0, col],
        "| Category:", header_preview.iloc[1, col],
        "| Field:", header_preview.iloc[2, col]
    )
```

```python
# Group residential electricity data by state
state_summary = electricity_clean.groupby("State").agg(
    Total_Revenue=("Revenue_Thousand_Dollars", "sum"),
    Total_Sales=("Sales_MWh", "sum"),
    Total_Customers=("Customers", "sum")
).reset_index()

# Calculate the state average residential price
state_summary["Price_Cents_kWh"] = (
    state_summary["Total_Revenue"]
    / state_summary["Total_Sales"]
) * 100

# Sort states from highest to lowest price
state_summary = state_summary.sort_values(
    by="Price_Cents_kWh",
    ascending=False
)

# Display the first 10 states
display(state_summary.head(10))
```

```python
import matplotlib.pyplot as plt

# Select the 10 states with the highest prices
top10 = state_summary.head(10)

# Create the bar chart
plt.figure(figsize=(10, 6))

plt.bar(
    top10["State"],
    top10["Price_Cents_kWh"]
)

plt.title("Top 10 States by Residential Electricity Price (2025)")
plt.xlabel("State")
plt.ylabel("Average Price (Cents per kWh)")

plt.tight_layout()
plt.show()
```

```python
# Step 38 - Data center electricity demand

import pandas as pd

data_center_demand = pd.DataFrame({
    "Year": [2024, 2028],
    "Electricity_TWh": [192, 464],
    "Data_Type": [
        "Historical Estimate",
        "Reference Forecast"
    ]
})

# Convert TWh to MWh
data_center_demand["Electricity_MWh"] = (
    data_center_demand["Electricity_TWh"] * 1_000_000
)

display(data_center_demand)
```

```python
# Step 39 - Calculate projected data center demand growth

demand_2024 = 192
demand_2028 = 464

# Calculate the increase in TWh
demand_increase = demand_2028 - demand_2024

# Calculate percentage growth
growth_percent = (
    demand_increase / demand_2024
) * 100

print("2024 electricity demand:", demand_2024, "TWh")
print("2028 projected demand:", demand_2028, "TWh")
print("Projected increase:", demand_increase, "TWh")
print("Projected percentage growth:", round(growth_percent, 2), "%")
```

```python
# Step 40 - Chart data center electricity demand

import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

bars = plt.bar(
    data_center_demand["Year"].astype(str),
    data_center_demand["Electricity_TWh"],
    color=["steelblue", "orange"]
)

# Add electricity demand values above the bars
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height + 8,
        f"{height:.0f} TWh",
        ha="center"
    )

plt.xlabel("Year")
plt.ylabel("Electricity Consumption (TWh)")
plt.title("U.S. Data Center Electricity Demand: 2024 vs. 2028")

plt.ylim(0, 550)
plt.tight_layout()
plt.show()
```

*Source: Lawrence Berkeley National Laboratory, United States Data Center Energy Usage Report: 2025 Update.*

```python
# Step 42 - Hypothetical utility infrastructure cost model

# Scenario assumptions
infrastructure_cost = 1_000_000_000
annual_recovery_rate = 0.10
residential_share = 0.20
residential_customers = 500_000

# Calculate annual infrastructure cost recovery
annual_cost = infrastructure_cost * annual_recovery_rate

# Calculate the portion assigned to residential customers
residential_cost = annual_cost * residential_share

# Calculate the average monthly cost per customer
monthly_bill_increase = (
    residential_cost / residential_customers / 12
)

print("Annual infrastructure cost: $", f"{annual_cost:,.2f}")
print("Residential cost share: $", f"{residential_cost:,.2f}")
print("Monthly increase per customer: $", f"{monthly_bill_increase:,.2f}")
```

```python
# Step 43 - Compare infrastructure cost-sharing scenarios

import pandas as pd

scenarios = pd.DataFrame({
    "Scenario": [
        "Higher Residential Allocation",
        "Moderate Residential Allocation",
        "Lower Residential Allocation"
    ],
    "Residential_Share": [0.30, 0.20, 0.10]
})

# Calculate annual costs assigned to residential customers
scenarios["Annual_Residential_Cost"] = (
    annual_cost * scenarios["Residential_Share"]
)

# Calculate monthly cost per residential customer
scenarios["Monthly_Bill_Increase"] = (
    scenarios["Annual_Residential_Cost"]
    / residential_customers
    / 12
)

display(scenarios.round(2))
```

```python
# Step 44 - Chart residential cost-sharing scenarios

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

bars = plt.bar(
    scenarios["Scenario"],
    scenarios["Monthly_Bill_Increase"],
    color=["firebrick", "orange", "seagreen"]
)

# Add dollar amounts above each bar
for bar in bars:
    height = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height + 0.1,
        f"${height:.2f}",
        ha="center"
    )

plt.ylabel("Monthly Bill Increase ($)")
plt.xlabel("Cost-Sharing Scenario")
plt.title("Hypothetical Residential Electricity Bill Impact")

plt.ylim(0, 6)
plt.xticks(rotation=15, ha="right")

plt.tight_layout()
plt.show()
```

### Infrastructure Cost Allocation and Residential Electricity Bills

**Analysis:**

I developed a hypothetical utility infrastructure cost model to examine how different cost-sharing arrangements could affect residential electricity bills.

The model assumes a $1 billion infrastructure investment, a 10% annual cost recovery rate, and 500,000 residential customers.

Under these assumptions:

- A 30% residential cost allocation results in a $5.00 monthly increase per customer.
- A 20% allocation results in a $3.33 monthly increase.
- A 10% allocation results in a $1.67 monthly increase.

Reducing the residential allocation from 30% to 10% decreases the modeled monthly impact by approximately 67%.

**Potential Strategies:**

Utilities and regulators could consider requiring large data center customers to contribute directly to infrastructure costs, establishing long-term service agreements, and designing electricity rates that better reflect the costs of serving large loads.

**Conclusion:**

The analysis illustrates that residential electricity bill impacts depend partly on how infrastructure investments are allocated among customer groups.

These results are hypothetical and should not be interpreted as actual utility rate increases or forecasts.

```python
# Step 49 - Long-term data center agreement scenario

# Hypothetical annual infrastructure costs
annual_infrastructure_cost = 100_000_000

# Data center payments
payment_without_agreement = 40_000_000
payment_with_agreement = 80_000_000

# Calculate remaining infrastructure costs
gap_without_agreement = (
    annual_infrastructure_cost - payment_without_agreement
)

gap_with_agreement = (
    annual_infrastructure_cost - payment_with_agreement
)

# Calculate reduction in funding gap
gap_reduction = (
    gap_without_agreement - gap_with_agreement
)

print("Without agreement - Funding gap: $",
      f"{gap_without_agreement:,.2f}")

print("With agreement - Funding gap: $",
      f"{gap_with_agreement:,.2f}")

print("Reduction in funding gap: $",
      f"{gap_reduction:,.2f}")
```

```python
# Step 50 - Residential impact of long-term agreements

residential_share = 0.20
residential_customers = 500_000

# Residential costs without agreement
residential_cost_without = (
    gap_without_agreement * residential_share
)

# Residential costs with agreement
residential_cost_with = (
    gap_with_agreement * residential_share
)

# Calculate monthly household charges
monthly_without = (
    residential_cost_without / residential_customers / 12
)

monthly_with = (
    residential_cost_with / residential_customers / 12
)

# Calculate avoided monthly charges
monthly_savings_agreement = monthly_without - monthly_with

print("Without agreement: $", f"{monthly_without:.2f}", "per month")
print("With agreement: $", f"{monthly_with:.2f}", "per month")
print("Avoided monthly charge: $", f"{monthly_savings_agreement:.2f}")
```

```python
# Step 51 - Chart residential impact of long-term agreements

import matplotlib.pyplot as plt

agreement_scenarios = [
    "Without Agreement",
    "With Agreement"
]

monthly_costs = [
    monthly_without,
    monthly_with
]

plt.figure(figsize=(8, 5))

bars = plt.bar(
    agreement_scenarios,
    monthly_costs,
    color=["firebrick", "seagreen"]
)

# Add dollar amounts above each bar
for bar in bars:
    height = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height + 0.05,
        f"${height:.2f}",
        ha="center"
    )

plt.ylabel("Monthly Infrastructure Charge ($)")
plt.xlabel("Data Center Service Agreement")
plt.title("Hypothetical Residential Impact of Long-Term Agreements")

plt.ylim(0, 2.5)
plt.tight_layout()
plt.show()
```

```python
# Step 53 - Flexible data center electricity demand

# Hypothetical assumptions
peak_demand_mw = 100
flexible_reduction = 0.20
annual_capacity_cost_per_mw = 150_000

# Calculate reduced peak demand
reduced_peak_mw = (
    peak_demand_mw * (1 - flexible_reduction)
)

# Calculate avoided peak capacity
avoided_capacity_mw = (
    peak_demand_mw - reduced_peak_mw
)

# Calculate potential annual capacity savings
annual_capacity_savings = (
    avoided_capacity_mw * annual_capacity_cost_per_mw
)

print("Original peak demand:", peak_demand_mw, "MW")
print("Reduced peak demand:", reduced_peak_mw, "MW")
print("Avoided peak capacity:", avoided_capacity_mw, "MW")
print("Potential annual capacity savings: $",
      f"{annual_capacity_savings:,.2f}")
```
